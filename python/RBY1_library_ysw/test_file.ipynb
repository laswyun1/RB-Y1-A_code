Dynamics Example 1
</span>

```python
# Create 2-Link system
link_0 = rby.dynamics.Link("link_0")
link_1 = rby.dynamics.Link("link_1")
joint_0 = rby.dynamics.Joint.make_revolute("joint_0", np.identity(4), np.array([0,0,1]))   # Z-axis revolute joint
joint_0.connect_links(link_0, link_1, np.identity(4), np.identity(4))
dyn_robot = rby.dynamics.Robot(rby.dynamics.RobotConfiguration(name="sample_robot", base_link=link_0))

dyn_state = dyn_robot.make_state(["link_0", "link_1"], ["joint_0"])
dyn_state.set_q(np.array([np.pi / 2]))  # Set angle of joint_0 as 90 degrees
```

<span style="color:#B0B0B0; font-weight:bold; font-size:30px;">
Dynamics Example 2
</span>

```python
# 1. Get the dynamics model (dyn_robot) from the robot.
dyn_robot = robot.get_dynamics()


# 2. Create a State object for dynamics calculations.
#    Specify the names of the links and joints to be used in the calculations.
dyn_state = dyn_robot.make_state(["base", "ee_right"], model.robot_joint_names)


# 3. Define and set the robot's state (q, qdot, qddot).
#    Here, we use random joint angles for demonstration.
q = (np.random.rand(model.robot_dof) - 0.5) * np.pi / 2
dyn_state.set_gravity(np.array([0, 0, 0, 0, 0, -9.81]))
dyn_state.set_q(q)
dyn_state.set_qdot(np.zeros(model.robot_dof))
dyn_state.set_qddot(np.zeros(model.robot_dof))

print(f"{dyn_state = }")


# 4. Perform Inverse Dynamics calculation.
#    To compute inverse dynamics, forward kinematics-related calculations must be performed first.
#    These functions cache their results within the dyn_state object.

# 4-1. Compute forward kinematics (FK) to get the position/orientation of each link.
dyn_robot.compute_forward_kinematics(dyn_state)

# 4-2. Compute the first derivative of FK (Jacobian) to get the velocity of each link.
#      Must be called after `compute_forward_kinematics`.
dyn_robot.compute_diff_forward_kinematics(dyn_state)

# 4-3. Compute the second derivative of FK to get the acceleration of each link.
#      Must be called after `compute_diff_forward_kinematics`.
dyn_robot.compute_2nd_diff_forward_kinematics(dyn_state)

# 4-4. Based on the kinematic information calculated above,
#      compute the joint torques (tau) required to maintain the current state (q, qdot, qddot).
dyn_robot.compute_inverse_dynamics(dyn_state)

# 5. Print the calculated torques.
with np.printoptions(precision=4, suppress=True):
    print(f"Inverse dynamics torque (Nm): {dyn_state.get_tau()}")
```

Recording
</span>

```python
### Record Data ###

# --- For .ipynb file --------------------------------------------------
import time

recorder = ysw.Recorder(robot, filename="recorded_data.npz", rate=10)
recorder.start_record()
time.sleep(10)
recorder.stop_record()


# --- For  .py file --------------------------------------------------
# recorder = ysw.Recorder(robot, filename="recorded_data.npz", rate=10)
# recorder.run_record()
```

```python
### Load recorded file and Replay ###

body_indices = model.torso_idx + model.right_arm_idx + model.left_arm_idx   # 20-DoF
body_joint_names = [model.robot_joint_names[i] for i in body_indices]
saved_traj = np.load("recorded_data.npz", allow_pickle=True)["data"]

stream = robot.create_command_stream(10)

for i, traj_step in enumerate(saved_traj):
    # First step: dt = 5, others: dt = 0.1
    dt = 0.1 if i > 0 else 5
    traj_step_body = traj_step[body_indices]
    rc = rby.RobotCommandBuilder().set_command(
            rby.ComponentBasedCommandBuilder().set_body_command(
                rby.JointPositionCommandBuilder().set_command_header(
                    rby.CommandHeaderBuilder().set_control_hold_time(1)
                )
                .set_minimum_time(dt)
                .set_position(traj_step_body)
            )
         )
    rv = stream.send_command(rc)

    time.sleep(dt * 0.95) 
```
